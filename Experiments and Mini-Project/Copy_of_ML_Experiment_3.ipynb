## Task 8: Standardize Numerical Features

Identify the numerical predictors that need to be standardized.

Use the following standardization equation:

\[
z = ($x$-$\mu$)/$\sigma$
\]

where:

- $x$ = original feature value
- $\mu$ = mean of the feature
- $\sigma$ = standard deviation of the feature

Perform the following:

- Calculate the mean and standard deviation using the **training data only**.
- Standardize the training features.
- Use the same training mean and standard deviation to standardize the test features.
- Do not standardize the target variable `Selling_Price`.

---

```python
# Task 8: Standardize Numerical Features

import numpy as np
import pandas as pd

# --------------------------------------------------
# Step 1: Create X and y
# --------------------------------------------------

# Predictor variables
# Correct 'selling_price' and 'car_name' casing.
# Drop 'Car_Age' and 'Unnamed: 0' here as they are redundant or non-predictive,
# causing multicollinearity or numerical instability.
X = df.drop(['selling_price', 'Year', 'car_name', 'Car_Age', 'Unnamed: 0'], axis=1, errors='ignore')

# Target variable
# Correct 'selling_price' casing
y = df['selling_price']

# --------------------------------------------------
# Step 2: Encode categorical variables
# --------------------------------------------------

# Correct casing for categorical columns and include 'brand' and 'model'
categorical_columns = ['brand', 'model', 'fuel_type', 'seller_type', 'transmission_type']

X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

# --------------------------------------------------
# Step 3: Split data into training and testing sets
# --------------------------------------------------

# Use 80% data for training and 20% for testing
np.random.seed(42)

indices = np.arange(len(X_encoded))
np.random.shuffle(indices)

split_index = int(0.80 * len(X_encoded))

train_indices = indices[:split_index]
test_indices = indices[split_index:]

X_train = X_encoded.iloc[train_indices].copy()
X_test = X_encoded.iloc[test_indices].copy()

y_train = y.iloc[train_indices].copy()
y_test = y.iloc[test_indices].copy()

# --------------------------------------------------
# Step 4: Identify numerical predictors
# --------------------------------------------------

# Correct numerical features based on existing columns in df and X_encoded
# 'Car_Age' and 'Unnamed: 0' removed as they are redundant or non-predictive
numerical_features = [
    'vehicle_age',
    'km_driven',
    'mileage',
    'engine',
    'max_power',
    'seats'
]

print("Numerical features to be standardized:")
print(numerical_features)

# --------------------------------------------------
# Step 5: Calculate mean and standard deviation
#         using TRAINING DATA ONLY
# --------------------------------------------------

train_mean = X_train[numerical_features].mean()
train_std = X_train[numerical_features].std()

print("\nTraining Mean:")
print(train_mean)

print("\nTraining Standard Deviation:")
print(train_std)

# --------------------------------------------------
# Step 6: Standardize training numerical features
# --------------------------------------------------

X_train[numerical_features] = (
    X_train[numerical_features] - train_mean
) / train_std

# --------------------------------------------------
# Step 7: Standardize test numerical features
#         using training mean and standard deviation
# --------------------------------------------------

X_test[numerical_features] = (
    X_test[numerical_features] - train_mean
) / train_std

# --------------------------------------------------
# Step 8: Remove zero-variance columns (multicollinearity fix)
# --------------------------------------------------
# Identify columns in X_train that have zero variance (i.e., are constant)
# Such columns cause multicollinearity when combined with the bias term.
zero_variance_cols = X_train.columns[X_train.var() == 0]

if len(zero_variance_cols) > 0:
    print(f"\nWarning: Found {len(zero_variance_cols)} zero-variance columns in X_train: {zero_variance_cols.tolist()}")
    print("These columns are constant in the training set and cause multicollinearity. Dropping them.")
    # Drop these columns from both training and testing sets
    X_train = X_train.drop(columns=zero_variance_cols)
    X_test = X_test.drop(columns=zero_variance_cols)
else:
    print("\nNo zero-variance columns found in X_train after one-hot encoding and standardization.")

# --------------------------------------------------
# Step 9: Verify the standardized training data
# --------------------------------------------------

print("\nStandardized Training Features:")
display(X_train.head())

# --------------------------------------------------
# Step 10: Verify the standardized test data
# --------------------------------------------------

print("\nStandardized Test Features:")
display(X_test.head())

# --------------------------------------------------
# Step 11: Verify that Selling_Price is NOT standardized
# --------------------------------------------------

print("\nTraining Target (Selling_Price):")
display(y_train.head())

print("\nTest Target (Selling_Price):")
display(y_test.head())

# --------------------------------------------------
# Step 12: Verify training means after standardization
# --------------------------------------------------

print("\nMean of standardized training features:")
print(X_train[numerical_features].mean())

print("\nStandard deviation of standardized training features:")
print(X_train[numerical_features].std())
```

## Task 12: Implement the OLS Function

Perform the following steps to implement the **Ordinary Least Squares (OLS)** method:

- Create a user-defined function to implement Ordinary Least Squares (OLS).
- Calculate the matrix $X^TX$.
- Calculate the matrix $X^T y$.
- Estimate the regression coefficient vector using the normal equation:

: β̂ = (XᵀX)⁻¹Xᵀy.

- Return the calculated coefficient vector.
- Use **NumPy matrix operations** to perform the calculations.
- Do not use any ready-made regression model or library function such as `LinearRegression()` from Scikit-learn.\

```python
# Task 12: Implement Ordinary Least Squares (OLS)

import numpy as np

# --------------------------------------------------
# User-defined OLS function
# --------------------------------------------------

def ols(X, y):
    """
    Calculate OLS regression coefficients using
    the normal equation:

        beta = (X^T X)^(-1) X^T y

    Pseudo-inverse is used to handle singular matrices.
    """

    # Calculate X transpose
    XT = X.T

    # Calculate X^T X
    XTX = XT @ X

    # Calculate X^T y
    XTy = XT @ y

    # Calculate pseudo-inverse of X^T X
    XTX_pinv = np.linalg.pinv(XTX)

    # Calculate regression coefficients
    beta = XTX_pinv @ XTy

    # Return coefficient vector
    return beta


# --------------------------------------------------
# Calculate OLS coefficients using training data
# --------------------------------------------------

beta = ols(Xtrain_bias, ytrain)

# Display coefficient vector
print("OLS Regression Coefficient Vector:")
print(beta)

# Display shape
print("\nShape of coefficient vector:")
print(beta.shape)

# --------------------------------------------------
# Display the matrices used in OLS
# --------------------------------------------------

XTX = Xtrain_bias.T @ Xtrain_bias
XTy = Xtrain_bias.T @ ytrain

print("\nX^T X Matrix:")
print(XTX)

print("\nX^T y Vector:")
print(XTy)

print("\nRegression Coefficient Vector (Beta):")
print(beta)
```

###Limitations of Multiple Linear Regression
